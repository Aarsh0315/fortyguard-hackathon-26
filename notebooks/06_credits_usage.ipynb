# 06 — Credits & Usage

Two endpoints for monitoring consumption:

- `POST /v1/system/fetch-api-key-usage` — current billing cycle summary
- `POST /v1/system/fetch-api-key-custom-usage` — usage over an arbitrary window

Credits are only charged on tasks that reach `succeeded`. Failed tasks are free.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))
from dotenv import load_dotenv; load_dotenv(pathlib.Path.cwd().parent / '.env')

from fortyguard import FortyGuardClient
client = FortyGuardClient()
```

```python
import pandas as pd
usage = client.fetch_api_key_usage()
print('Plan      :', usage.get('plan_details', {}).get('plan_type'))
print('Period    :', usage.get('plan_details', {}).get('billing_period'))
print('Used      :', usage.get('credit_summary', {}).get('cycle_credits_used'))
print('Remaining :', usage.get('credit_summary', {}).get('cycle_remaining_credits'))

breakdown = pd.DataFrame(usage.get('activity_breakdown', []))
breakdown
```

```python
import matplotlib.pyplot as plt
if not breakdown.empty:
    breakdown.plot(kind='barh', x='name', y='credits', legend=False, color='steelblue', figsize=(8, 3))
    plt.xlabel('Credits used'); plt.title('Usage by activity (this cycle)'); plt.tight_layout(); plt.show()
```

```python
# Custom date range. Dates can be YYYY-MM-DD — the client formats them to ISO for you.
custom = client.fetch_api_key_custom_usage(start_date='2024-07-01', end_date='2024-07-31')
print('Range       :', custom.get('date_range', {}).get('date_range_formatted'))
print('Total used  :', custom.get('total_credits_used'))
pd.DataFrame(custom.get('activity_breakdown', []))
```
